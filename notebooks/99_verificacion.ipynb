```python
P = "workspace.nyctaxi_project"

n_origen = spark.table("samples.nyctaxi.trips").count()
n_bronze = spark.table(f"{P}.bronze_trips").count()
n_silver = spark.table(f"{P}.silver_trips").count()
n_rej    = spark.table(f"{P}.silver_rejected_trips").count()

n_gold_hour = spark.sql(f"SELECT SUM(trips) FROM {P}.gold_demand_by_hour").first()[0]
n_gold_zone = spark.sql(f"SELECT SUM(trips) FROM {P}.gold_revenue_by_zone").first()[0]
n_gold_dist = spark.sql(f"SELECT SUM(trips) FROM {P}.gold_fare_by_distance").first()[0]
n_gold_day  = spark.sql(f"SELECT SUM(trips) FROM {P}.gold_daily_kpis").first()[0]

pct_rej = 100 * n_rej / n_bronze

print(f"Origen: {n_origen} | Bronze: {n_bronze} | Silver: {n_silver} | Rechazados: {n_rej} ({pct_rej:.2f}%)")

assert n_bronze <= n_origen, "Bronze tiene más filas que el origen (duplicados)"
assert n_silver + n_rej == n_bronze, "Silver + Rechazados no coincide con Bronze"
assert n_gold_hour == n_gold_zone == n_gold_dist == n_gold_day == n_silver, "Gold no coincide con Silver"
assert pct_rej < 5, f"Porcentaje de rechazo anormal: {pct_rej:.2f}%"

print("Todos los chequeos de calidad pasaron")
```
